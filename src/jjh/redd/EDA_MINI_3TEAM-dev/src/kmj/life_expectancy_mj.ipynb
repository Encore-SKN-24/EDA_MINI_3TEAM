```python
df_alcohol = pd.read_csv('./data/alcohol_raw_data.csv')
df_alcohol.head()
```

```python
df_alcohol.columns.to_list()
# 표 분석
# ParentLocationCode = 지역코드
# ParentLocation = 지역
# SpatialDimValueCode = 국가코드
# Location = 국가
# Period = Year
# FactValueNumeric = 순수 알콜 값(=> pure_alcohol), Value = 소수점 둘째자리에서 반올림
# 수치가 10 이면 알코올을 진짜 10리터 마셨다 -> 4도짜리면 25배, 20도 짜리면 5배 하면 실제값 추측 가능
```

```python
# 나라 이름 맞춰주기
df['Country'] = df['Country'].replace({
    "Côte d'Ivoire" : "Cote d'Ivoire" 
})


df_alcohol['Location'] = df_alcohol['Location'].replace({
    "Netherlands (Kingdom of the)": "Netherlands",
    "Türkiye": "Turkey",
    "North Macedonia": "The former Yugoslav republic of Macedonia",
    "Eswatini": "Swaziland"
})

# Merge 실행
df_1 = pd.merge(
    df, 
    df_alcohol[['Location', 'Period', 'ParentLocation', 'SpatialDimValueCode', 'FactValueNumeric']], 
    left_on=['Country', 'Year'], 
    right_on=['Location', 'Period'], 
    how='left'
)

# [4] 컬럼명 정리 및 중복 제거
df_1 = df_1.rename(columns={
    'ParentLocation': 'Region',
    'SpatialDimValueCode': 'Iso_code',
    'FactValueNumeric': 'Alcohol_Consumption'
}).drop(columns=['Location', 'Period'])
```

```python
df_1.isna().sum()
```

```python
cols = df_1.columns.to_list()
cols.remove('Region')
cols.remove('Iso_code')
cols.insert(0, 'Region')
cols.insert(1, 'Iso_code')
df_1 = df_1[cols]
df_1 = df_1.drop('Alcohol', axis=1)
```

```python
df_1 = df_1.drop('Population', axis=1)
```

```python
df_1 = df_1.drop('Income composition of resources', axis=1)
```

```python
df_1.isna().sum()
```

```python
# 나라 찾기 -> csv가 유럽문자 때문에 안불러와져서 따로 확인해봄 (코트디부아르, 튀르키예)
t_countries = df_1[df_1['Country'].str.startswith('T', na=False)]['Country'].unique()
print(t_countries)
```

```python
# schooling data 가져오기
df_schooling = pd.read_csv('./data/mean_year_schooling.csv')
df_schooling.head()
```

```python
# 데이터 분석
# mys_연도 -> 연도별 데이터
# 데이터 구조가 조금 이상한데...
```

```python
mys_columns_df = df_schooling.filter(regex='^mys')
mys_columns_df["Iso_code"] = df_schooling["iso3"]
cols = mys_columns_df.columns.to_list()
cols = cols[:34]
cols.insert(0, "Iso_code")
mys_columns_df = mys_columns_df[cols]
mys_columns_df
```

```python
df_mys = mys_columns_df.melt(
    id_vars=["Iso_code"],
    value_vars=mys_columns_df.filter(like='mys').columns,
    var_name="Year",
    value_name = "Schooling"
)

df_mys['Year'] = df_mys['Year'].str.replace('mys_','').astype(int)
df_mys
```

```python
# 데이터 구해왔으니 합치기

df_2 = pd.merge(
    df_1, 
    df_mys,
    on=['Iso_code', 'Year'], 
    how='left'
)

df_2
```

```python
df_2 = df_2.drop("SchooX1ling", axis=1)
df_2.isna().sum()
```

```python
# 국가별로 schooling 칼럼의 결측치 개수 확인
missing_pattern = df_2.groupby('Iso_code')['Schooling'].apply(lambda x: x.isnull().sum())

# 결측치가 1개라도 있는 나라만 출력
print(missing_pattern[missing_pattern > 0])
```

```python
df_2 = df_2.sort_values(['Iso_code', 'Year'])
df_2['Schooling'] = df_2.groupby('Iso_code')['Schooling'].transform(lambda x: x.interpolate(method='linear'))
df_3 = df_2.dropna(subset=['Schooling'])
```
